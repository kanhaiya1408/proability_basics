# Probability Basics Assignment Answers

### **Question 1: A die is rolled. What is the probability of getting:**
**(a) An even number**
* **Solution:**
  * Sample space $S = \{1, 2, 3, 4, 5, 6\} \implies n(S) = 6$
  * Even numbers are $\{2, 4, 6\} \implies$ number of favorable outcomes = 3.
  * Probability = $\frac{3}{6} = \frac{1}{2} = 0.5$ (or 50%).

**(b) A number greater than 4**
* **Solution:**
  * Numbers greater than 4 are $\{5, 6\} \implies$ number of favorable outcomes = 2.
  * Probability = $\frac{2}{6} = \frac{1}{3} \approx 0.333$ (or 33.3%).

### **Question 2: In a class of 50 students, 20 like Mathematics (M), 15 like Science (S), and 5 like both subjects. What is the probability that a student chosen at random likes Mathematics or Science?**
* **Solution:**
  * Total students, $n(U) = 50$
  * $P(M) = \frac{20}{50} = 0.4$
  * $P(S) = \frac{15}{50} = 0.3$
  * $P(M \cap S) = \frac{5}{50} = 0.1$
  * We need the probability of liking Math OR Science ($P(M \cup S)$):
    $$P(M \cup S) = P(M) + P(S) - P(M \cap S) = 0.4 + 0.3 - 0.1 = 0.6$$
* **Answer:** **0.6 (or 60%)**

---

### **Question 3: A bag has 3 red and 2 blue balls. If one ball is drawn randomly and is red, what is the probability that the next ball is also red (without replacement)?**
* **Solution:**
  * Initial: 3 Red, 2 Blue (Total = 5 balls)
  * After drawing 1 Red ball, we do not replace it:
    * Remaining Red balls = $3 - 1 = 2$
    * Remaining Blue balls = 2
    * Total remaining balls = 4
  * Probability of the next ball being Red = $\frac{\text{Remaining Red}}{\text{Total Remaining}} = \frac{2}{4} = \frac{1}{2} = 0.5$.
* **Answer:** **0.5 (or 50%)**

---

### **Question 4: The population of a school is divided into 60% boys and 40% girls. If you want equal representation of both genders in the sample, which method should you use: Simple Random Sampling or Stratified Sampling? Why?**
* **Solution/Answer:**
  * You should use **Stratified Sampling**.
  * **Why:** In Stratified Sampling, the population is divided into subgroups (strata) based on specific characteristics (in this case, gender: Boys and Girls). To ensure equal representation of both genders in the sample, we can deliberately select an equal number of respondents from each stratum (e.g., 50 boys and 50 girls). Simple Random Sampling relies purely on chance, which may lead to an unequal distribution of genders in the final sample.

### **Question 5: The average height of 1000 students = 160 cm. A sample of 100 students shows an average height = 158 cm. Find the sampling error.**
* **Solution:**
  * Population Mean ($\mu$) = 160 cm
  * Sample Mean ($\bar{x}$) = 158 cm
  * $\text{Sampling Error} = \bar{x} - \mu = 158 - 160 = -2\text{ cm}$ (or an absolute error of $2\text{ cm}$).
* **Answer:** **-2 cm**

---

### **Question 6: The population mean salary is ₹50,000 with $\sigma$ = ₹5,000. If we take a sample of 100 employees, what is the standard error of the mean (SEM)?**
* **Solution:**
  * Standard Deviation ($\sigma$) = 5000
  * Sample size ($n$) = 100
  * Formula for Standard Error of Mean (SEM):
    $$\text{SEM} = \frac{\sigma}{\sqrt{n}} = \frac{5000}{\sqrt{100}} = \frac{5000}{10} = 500$$
* **Answer:** **₹500**

---

### **Question 7: In a group of 100 students, 40 like Cricket (C), 30 like Football (F), and 10 like both Cricket and Football. Find the probability that a student likes at least one sport.**
* **Solution:**
  * Total students = 100
  * $P(C) = \frac{40}{100} = 0.4$
  * $P(F) = \frac{30}{100} = 0.3$
  * $P(C \cap F) = \frac{10}{100} = 0.1$
  * Probability of liking at least one sport ($P(C \cup F)$):
    $$P(C \cup F) = P(C) + P(F) - P(C \cap F) = 0.4 + 0.3 - 0.1 = 0.6$$
* **Answer:** **0.6 (or 60%)**

---

### **Question 8: From a deck of 52 cards, two cards are drawn without replacement. What is the probability that both are Aces?**
* **Solution:**
  * Total Aces in a deck = 4
  * Probability of 1st card being an Ace = $\frac{4}{52}$
  * After drawing one Ace without replacement:
    * Remaining Aces = 3
    * Remaining Cards = 51
  * Probability of 2nd card being an Ace = $\frac{3}{51}$
  * Probability of both being Aces:
    $$P(\text{Both Aces}) = \frac{4}{52} \times \frac{3}{51} = \frac{1}{13} \times \frac{1}{17} = \frac{1}{221} \approx 0.0045$$
* **Answer:** **$\frac{1}{221}$ (or $\approx 0.45\%$)**

---

### **Question 9: A factory produces bulbs with 2% defective rate. If 5 bulbs are chosen at random, what is the probability that all are non-defective?**
* **Solution:**
  * Probability of a bulb being defective ($p$) = 0.02
  * Probability of a bulb being non-defective ($q$) = $1 - 0.02 = 0.98$
  * Number of trials ($n$) = 5
  * Probability that all 5 are non-defective:
    $$P(\text{All non-defective}) = q^5 = (0.98)^5 \approx 0.9039$$
* **Answer:** **$\approx 0.9039$ (or 90.39%)**

### **Question 10: Differentiate between discrete and continuous random variables with examples.**
* **Solution:**
  
| Feature | Discrete Random Variable | Continuous Random Variable |
| :--- | :--- | :--- |
| **Definition** | Variables that can take only countable, distinct values. | Variables that can take any value within a continuous range or interval. |
| **Nature of values** | Integers, whole numbers (no fractions between two successive values). | Real numbers (includes decimals and fractions). |
| **Examples** | 1. Number of heads in 3 coin tosses.<br>2. Number of children in a family.<br>3. Number of cars in a parking lot. | 1. Height or weight of students.<br>2. Temperature of a room.<br>3. Time taken to complete a test. |

```python
# Question 9 Calculation
# A factory produces bulbs with 2% defective rate.
# If 5 bulbs are chosen at random, find the probability that all are non-defective.

def_rate = 0.02
non_def_rate = 1 - def_rate
n_bulbs = 5

# Probability that all 5 bulbs are non-defective
prob_all_non_defective = non_def_rate ** n_bulbs

print(f"Defective Rate: {def_rate * 100}%")
print(f"Non-defective Rate: {non_def_rate * 100}%")
print(f"Probability that all {n_bulbs} bulbs are non-defective: {prob_all_non_defective:.6f} (or {prob_all_non_defective * 100:.4f}%)")
```

```python
# Question 8 Calculation
# From a deck of 52 cards, two cards are drawn without replacement.
# What is the probability that both are Aces?

# Standard deck setup
total_cards = 52
total_aces = 4

# First draw
p_first_ace = total_aces / total_cards

# Second draw without replacement
p_second_ace_given_first = (total_aces - 1) / (total_cards - 1)

# Joint probability
p_both_aces = p_first_ace * p_second_ace_given_first

print(f"Probability of first card being an Ace: {p_first_ace:.4f} ({total_aces}/{total_cards})")
print(f"Probability of second card being an Ace: {p_second_ace_given_first:.4f} ({total_aces - 1}/{total_cards - 1})")
print(f"Probability that both are Aces: {p_both_aces:.6f} (or {p_both_aces * 100:.4f}%)")
print(f"Exact fraction representation: 1/{int(1/p_both_aces)}")
```
